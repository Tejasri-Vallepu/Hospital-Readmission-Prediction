## 2.Data Cleaning and Feature Selection

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/diabetic_data.csv")

print("Shape:", df.shape)
```

```python
df["readmitted"].value_counts()
```

```python
df["readmitted_30"] = (
    df["readmitted"] == "<30"
).astype(int)
```

```python
df = df.drop(columns=["readmitted"])
```

```python
columns_to_drop = [
    "weight",
    "medical_specialty",
    "payer_code"
]

df = df.drop(columns=columns_to_drop)
```

```python
columns_to_fill = [
    "race",
    "diag_1",
    "diag_2",
    "diag_3"
]

for column in columns_to_fill:
    df[column] = df[column].fillna("Unknown")
```

```python
missing_values = df.isnull().sum()

print(
    missing_values[missing_values > 0]
)
```

```python
print("Shape after cleaning:", df.shape)
```

```python
df = df.drop(
    columns=["encounter_id", "patient_nbr"]
)
```

```python
categorical_columns = df.select_dtypes(
    include=["object"]
).columns

for column in categorical_columns:
    print(column)
    print(df[column].nunique())
```

```python
constant_columns = [
    column
    for column in df.columns
    if df[column].nunique(dropna=False) <= 1
]

print("Constant columns:", constant_columns)
```

```python
df = df.drop(columns=constant_columns)
```

```python
print(df.dtypes)
```

```python
X = df.drop(columns=["readmitted_30"])
y = df["readmitted_30"]
```

```python
print(y.value_counts())
```

```python
print(
    y.value_counts(normalize=True) * 100
)
```

```python
numeric_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
potential_leakage = [
    "discharge_disposition_id",
    "admission_type_id",
    "admission_source_id"
]

for column in potential_leakage:
    if column in df.columns:
        print(column, df[column].nunique())
```

```python
for column in X.columns:
    unique_values = X[column].nunique()
    
    if unique_values == 1:
        print("Constant:", column)
```

```python
print("Total rows:", len(df))
print("Unique patients:", df["patient_nbr"].nunique() if "patient_nbr" in df.columns else "ID removed")
```

```python
df.to_csv(
    "../data/diabetic_data_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully.")
```

```python
print("Final dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nTarget distribution:")
print(df["readmitted_30"].value_counts())

print("\nData types:")
print(df.dtypes.value_counts())
```
